```python
import sqlite3
import pandas as pd


table = 'bronze_income_stmt_quarterly'

with sqlite3.connect('../data/tase_data_warehouse.db') as conn:
    df = pd.read_sql(f"SELECT * FROM {table} ", conn)

df['date'] = pd.to_datetime(df['report_date'], utc=True).dt.date

df['net_income_matrix'] = (df['Net Income'] / 1e9).round(2)
df['Revenue_matrix'] = (df['Total Revenue'] / 1e9).round(2)
df =df.rename(columns={'Diluted EPS': 'EPS_margin'}) 
df =df[['date','Revenue_matrix','net_income_matrix','EPS_margin','ticker']].sort_values(['ticker','date'], ascending=False)

df
```
